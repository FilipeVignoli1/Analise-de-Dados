```python
import duckdb # Importa o DuckDB para persistência dos dados
import pandas as pd # Importa o Pandas para tratamento dos arquivos

def importar_multiplos_csv_upsert(): # Função principal de carga
    conn = duckdb.connect('meu_projeto.db') # Conecta ao arquivo de banco de dados
    
    arquivos_para_tabelas = {
        'dados/controle de pessoa.csv': {'tabela': 'c_pessoa', 'pk': 'MAT.', 'sep': ';'},
        'dados/tipatend101.csv':        {'tabela': '101', 'pk': 'No.Atendimento', 'sep': ';'},
        'dados/tipatend1144.csv':       {'tabela': '1144', 'pk': 'No.Atendimento', 'sep': ';'},
        'dados/tipatend1763.csv':       {'tabela': '1763', 'pk': 'No.Atendimento', 'sep': ';'},
        'dados/tipatend1976.csv':       {'tabela': '1976', 'pk': 'No.Atendimento', 'sep': ';'},
        'dados/WF.csv':                 {'tabela': 'wf', 'pk': 'Protocolo_ID', 'sep': ';'}
    }

    for nome_arquivo, config in arquivos_para_tabelas.items():
        try:
            nome_tabela = config['tabela']
            chave_primaria = config['pk']
            separador = config['sep']

            with open(nome_arquivo, 'r', encoding='latin1', errors='ignore') as f:
                linhas_puladas = 0
                for linha in f:
                    # Busca a linha que contém a PK e tem estrutura de colunas
                    if chave_primaria in linha and linha.count(separador) > 5:
                        break
                    linhas_puladas += 1
                
                # Extrai e limpa os nomes das colunas da linha encontrada
                colunas_originais = [c.strip().replace('"', '') for c in linha.strip().split(separador)]
                
                # Lógica para renomear duplicados (Hiper Foco no erro Duplicate Names)
                colunas_unicas = []
                contagem = {}
                for col in colunas_originais:
                    if col in contagem:
                        contagem[col] += 1
                        colunas_unicas.append(f"{col}_{contagem[col]}")
                    else:
                        contagem[col] = 0
                        colunas_unicas.append(col)

                # Lê o restante do arquivo usando os novos nomes únicos de colunas
                df_temp = pd.read_csv(
                    f, 
                    sep=separador, 
                    names=colunas_unicas, 
                    on_bad_lines='skip', 
                    low_memory=False
                )

            # Garante que a PK no DataFrame não tenha espaços invisíveis
            if chave_primaria not in df_temp.columns:
                 # Caso a PK tenha sido renomeada por ser duplicada, ajustamos a referência
                 chave_primaria = [c for c in df_temp.columns if c.startswith(config['pk'])][0]

            # Processo de Upsert no DuckDB
            conn.execute("CREATE OR REPLACE TEMP TABLE tabela_temporaria AS SELECT * FROM df_temp")
            conn.execute(f'CREATE TABLE IF NOT EXISTS "{nome_tabela}" AS SELECT * FROM tabela_temporaria WHERE 1=0')
            conn.execute(f'DELETE FROM "{nome_tabela}" WHERE "{chave_primaria}" IN (SELECT "{chave_primaria}" FROM tabela_temporaria)')
            conn.execute(f'INSERT INTO "{nome_tabela}" SELECT * FROM tabela_temporaria')

            print(f"  Sucesso: Upsert na tabela '{nome_tabela}' — {len(df_temp)} linhas processadas.")

        except Exception as e:
            print(f"  Erro crítico em '{nome_arquivo}': {e}")

    conn.close() # Salva e fecha o banco

importar_multiplos_csv_upsert()
```
